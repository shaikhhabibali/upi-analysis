```python
all_data = []

for file in files:
    df = pd.read_excel(file)

    # Keep source file for data provenance
    df["Source File"] = file.name

    all_data.append(df)

upi = pd.concat(all_data, ignore_index=True)

print("Total rows:", len(upi))
print("Total columns:", len(upi.columns))
print("\nColumns:")
print(upi.columns.tolist())
```

```python
upi["Date"] = pd.to_datetime(
    upi["Month"],
    format="%B-%Y"
)

upi = upi.sort_values("Date").reset_index(drop=True)

print(upi[["Month", "Date"]].head())
print()
print(upi[["Month", "Date"]].tail())
```

```python
expected = pd.date_range(upi["Date"].min(), upi["Date"].max(), freq="MS")
missing = expected.difference(upi["Date"])

print("Expected months:", len(expected))
print("Months we have:", len(upi))
print("Missing months:", len(missing))
print(list(missing.strftime("%b-%Y")))
```

```python
print("Rows:", len(upi))
print("Missing values:", upi.isna().sum().sum())
print("Duplicate dates:", upi["Date"].duplicated().sum())
print("Date range:", upi["Date"].min(), "to", upi["Date"].max())
```

```python
upi = upi.rename(columns={
    "Volume (In Mn.)": "volume_mn",
    "Value (In Cr.)": "value_cr",
    "Source File": "source_file"
})

print(upi.columns.tolist())
```

```python
upi = upi.rename(columns={
    "Month": "month"
})

print(upi.columns.tolist())
```

```python
upi = upi.rename(columns={
    "Date": "date"
})

print(upi.columns.tolist())
```

```python
upi["financial_year"] = (
    upi["date"].dt.year
    - (upi["date"].dt.month < 4).astype(int)
).astype(str) + "-" + (
    (
        upi["date"].dt.year
        - (upi["date"].dt.month < 4).astype(int)
        + 1
    ).astype(str).str[-2:]
)

print(upi[["date", "financial_year"]].head(15))
```

```python
upi["month_number"] = upi["date"].dt.month
upi["month_name"] = upi["date"].dt.month_name()

print(upi[["date", "month_number", "month_name"]].head(15))
```

```python
# Convert transaction volume and value columns from text to numeric values for analysis

upi["volume_mn"] = (
    upi["volume_mn"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("\t", "", regex=False)
    .str.strip()
    .astype(float)
)

upi["value_cr"] = (
    upi["value_cr"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("\t", "", regex=False)
    .str.strip()
    .astype(float)
)

print(upi[["volume_mn", "value_cr"]].dtypes)
print(upi[["volume_mn", "value_cr"]].head())
```

```python
upi["avg_ticket_size"] = (
    upi["value_cr"] / upi["volume_mn"]
) * 10

print(
    upi[
        [
            "month",
            "volume_mn",
            "value_cr",
            "avg_ticket_size"
        ]
    ].head(15)
)
```

```python
# Calculate Month-over-Month (MoM) growth to measure the monthly change in UPI transaction volume

upi["mom_growth_pct"] = (
    upi["volume_mn"].pct_change() * 100
)

print(
    upi[
        [
            "month",
            "volume_mn",
            "mom_growth_pct"
        ]
    ].head(15)
)
```

```python
# Calculate Year-over-Year (YoY) growth to compare UPI transaction volume with the same month of the previous year

upi["yoy_growth_pct"] = (
    upi["volume_mn"].pct_change(periods=12) * 100
)

print(
    upi[
        [
            "month",
            "volume_mn",
            "yoy_growth_pct"
        ]
    ].head(20)
)
```

```python
# Validate the cleaned dataset to ensure there are no missing values or duplicate dates

print("Rows:", len(upi))
print("Columns:", len(upi.columns))
print("Missing values:", upi.isna().sum().sum())
print("Duplicate dates:", upi["date"].duplicated().sum())
print("Date range:", upi["date"].min(), "to", upi["date"].max())
```

```python
# Check which columns contain missing values after calculating YoY growth

print(upi.isna().sum())
```

```python
# Calculate MoM and YoY growth to measure changes in the total value of UPI transactions

upi["value_mom_growth_pct"] = (
    upi["value_cr"].pct_change() * 100
)

upi["value_yoy_growth_pct"] = (
    upi["value_cr"].pct_change(periods=12) * 100
)

print(
    upi[
        [
            "month",
            "value_cr",
            "value_mom_growth_pct",
            "value_yoy_growth_pct"
        ]
    ].head(20)
)
```

```python
# Perform a final quality check before saving the cleaned UPI dataset

print("Rows:", len(upi))
print("Columns:", len(upi.columns))

print("\nColumn names:")
print(upi.columns.tolist())

print("\nMissing values:")
print(upi.isna().sum())

print("\nDuplicate dates:", upi["date"].duplicated().sum())

print("\nDate range:")
print(upi["date"].min(), "to", upi["date"].max())
```

```python
# Save the cleaned and transformed UPI dataset for further analysis

clean_path = "../data/clean/upi_clean.csv"

upi.to_csv(clean_path, index=False)

print("Clean dataset saved successfully.")
print("File:", clean_path)
print("Rows:", len(upi))
print("Columns:", len(upi.columns))
```

```python
# Validate the final cleaned dataset and confirm that only expected growth values are missing

print("Rows:", len(upi))
print("Columns:", len(upi.columns))

print("\nMissing values:")
print(upi.isna().sum())

print("\nDuplicate dates:", upi["date"].duplicated().sum())

print("\nDate range:")
print(upi["date"].min(), "to", upi["date"].max())
```

```python
# Add SQL-friendly names, daily-average metrics, and financial-year completeness checks

upi = upi.rename(columns={
    "No. of Banks live on UPI": "banks_live"
})

# Calculate the number of days in each month
upi["days_in_month"] = upi["date"].dt.days_in_month

# Calculate daily average transaction volume and value
upi["avg_daily_volume_mn"] = (
    upi["volume_mn"] / upi["days_in_month"]
)

upi["avg_daily_value_cr"] = (
    upi["value_cr"] / upi["days_in_month"]
)

# Count how many months are available in each financial year
upi["fy_months_available"] = (
    upi.groupby("financial_year")["date"].transform("count")
)

# Validate date ordering and uniqueness
assert upi["date"].is_monotonic_increasing
assert upi["date"].is_unique

# Validate that transaction volume and value are positive
assert (upi[["volume_mn", "value_cr"]] > 0).all().all()

# Validate that there are no missing monthly periods
assert len(
    pd.date_range(
        upi["date"].min(),
        upi["date"].max(),
        freq="MS"
    )
) == len(upi)

# Save the updated clean dataset
upi.to_csv("../data/clean/upi_clean.csv", index=False)

print("Shape:", upi.shape)

print("\nSample:")
print(
    upi[
        ["month", "avg_daily_volume_mn", "avg_daily_value_cr",
         "fy_months_available"]
    ].head(3)
)

print("\nFinancial-year month counts:")
print(
    upi.groupby("financial_year")["date"].count()
)
```

```python
# Perform the final quality check after adding daily-average and financial-year metrics

print("Rows:", len(upi))
print("Columns:", len(upi.columns))

print("\nColumn names:")
print(upi.columns.tolist())

print("\nMissing values:")
print(upi.isna().sum())

print("\nDuplicate dates:", upi["date"].duplicated().sum())

print("\nDate range:")
print(upi["date"].min(), "to", upi["date"].max())

print("\nFinancial-year completeness:")
print(
    upi.groupby("financial_year")["date"].count()
)
```
